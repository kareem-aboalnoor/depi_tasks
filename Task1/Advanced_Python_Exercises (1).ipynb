## 9. Inheritance — Task
Create GraduateStudent.

```python
# Base class
class Student:
    def __init__(self, name, age, student_id):
        self.name = name
        self.age = age
        self.student_id = student_id

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"ID: {self.student_id}")


# Derived class (inherits from Student)
class GraduateStudent(Student):
    def __init__(self, name, age, student_id, research_area):

        super().__init__(name, age, student_id)
        self.research_area = research_area

    def display_info(self):
        
        super().display_info()
        print(f"Research Area: {self.research_area}")

        
grad_student = GraduateStudent("Kareem", 25, "G456", "Machine Learning in Embedded Systems")
grad_student.display_info()
```

## 10. Decorators — Task
Logger decorator.

```python
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Running {func.__name__}...")
        return func(*args, **kwargs)
    return wrapper

@logger
def say_hello(name):
    return f"Hello, {name}!"

# Example usage
print(say_hello("Kareem"))
```

## 11. Generators — Task
Countdown generator.

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1
# Example usage
for number in countdown(5):
    print(number)
```

## 12. Lambda, Map, Filter, Reduce — Task
Work with numbers list.

```python
from functools import reduce

lst = [1, 3, 6, 8, 2, 9]

# Map --> square each number
squares = list(map(lambda a: a**2, lst))
print("Squares:", squares)

# Filter --> keep only even numbers
evens = list(filter(lambda a: a % 2 == 0, lst))
print("Evens:", evens)

# Reduce --> sum all numbers
total = reduce(lambda a, b: a + b, lst)
print("Sum:", total)
```

```python
class Agent:
    def __init__(self, name, specialty):
        self.name = name
        self.specialty = specialty

    def report(self, mission):
        print(f"Agent {self.name} is assigned to {mission} using {self.specialty} skills.")

a = Agent("Echo", "Cybersecurity")
a.report("Network Defense")
```

```python
def divisible_by_three(n):
    for i in range(1, n+1):
        if i % 3 == 0:
            yield i
for num in divisible_by_three(20):
    print(num)
```
